````python
from typing import Literal
from pydantic import BaseModel, Field

class Sentiment(BaseModel):
    """The sentiment of the text."""
    sentiment: Literal["positive", "neutral", "negative"] = Field(..., description="The sentiment of the text")


model_with_structure = nvidia_model_1.with_structured_output(Sentiment)
````

### with invoke()

````python
response = model_with_structure.invoke("Kindness creates lasting joy.")
````

````python
print(response)
````

### with batch()

````python
response = model_with_structure.batch([
    "Kindness creates lasting joy.",
    "Success rewards persistent effort.",
    "I love Sunlight. It warms the skin.",

    "Pessemestic all the time.",
    "The storm caused damage!",

    "The clock ticks steadily."
])
````

````python
response
````

````python
from typing import Literal
from pydantic import BaseModel, Field
````

````python
class Categorization(BaseModel):
    """Multi-output categories"""
    categories: list[Literal["cars", "shopping", "sports", "study", "work"]] = Field(..., description="Multi-output categories")

model_with_structure = nvidia_model_1.with_structured_output(Categorization)
````

````python
response =  model_with_structure.batch([
     "That restoration looks incredible; you have a real talent for mechanics."
     "I found the perfect gift today! The staff was incredibly helpful."
     "Great game today! Your teamwork was the key to that victory."
     "Learning together helped me finally grasp these concepts. Thank you!"
 ])
````

````python
response
````

## Task 4

````python

class CandidateProfile(BaseModel):
    """Candidate profile details."""
    candidate_name: str = Field(..., description="The name of the candidate")
    skills: list[str] = Field(..., description="A list of skills the candidate possesses")
    experience_in_years: float = Field(..., description="Candidate experience in years")
````

````python

# Augment the LLM with schema for structured output
structured_llm_candidate = nvidia_model_1.with_structured_output(CandidateProfile)

text_input = """
Sarah Abdullah is a software engineer with expertise in Python, Java, and cloud computing.
she has 3 years of experience in developing scalable web applications and leading small teams.
"""

output = structured_llm_candidate.invoke(text_input)

print("Candidate Name:", output.candidate_name)
print("Skills:", output.skills)
print("Experience:", output.experience_in_years)
````

## Task 5

````python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        raise ValueError("Cannot divide by zero")
    return a / b
````

````python
from typing import Dict, Any

class ToolNameAndArguments(BaseModel):
    tool_name: str = Field(description="The name of the function to use")
    arguments: Dict[str, Any] = Field(description="The parameters to pass")

tool_calling_llm = nvidia_model_1.with_structured_output(ToolNameAndArguments)
````

````python
system_prompt = """
You are a math assistant. You have access to these tools:

def add(a, b)
def subtract(a, b)
def multiply(a, b)
def divide(a, b)

Return the tool name and arguments required to solve the user's request.
"""
````

````python
from langchain.messages import (
    SystemMessage,
    HumanMessage,
    AIMessage
)

response = tool_calling_llm.invoke([
    SystemMessage(system_prompt),
    HumanMessage("what is ten plus 100")
])
response
````

````python
if response.tool_name == 'add':
    result = add(response.arguments['a'], response.arguments['b'])

elif response.tool_name == 'subtract':
    result = subtract(response.arguments['a'], response.arguments['b'])

elif response.tool_name == 'multiply':
    result = multiply(response.arguments['a'], response.arguments['b'])

elif response.tool_name == 'divide':
    result = divide(response.arguments['a'], response.arguments['b'])
````
