```python
import pandas as pd
import sklearn
```

### Step 1. Load the data

```python
df = pd.read_csv("https://raw.githubusercontent.com/HassanAlgoz/B5/main/content/datasets/Rounded_Student_Hours_Studied_vs_Marks_Dataset.csv")
df.head()
```

### Step 2.a Assign variables `X` to the features and `y` to the target

```python
x = df['Hours_Studied']
y = df['Marks']
```

### Step 2.b print the type of each

```python
x.dtype, y.dtype
```

### Step 2.c identify whether the target is categorical or numerical & whether the task is regression or classification

```python
y.info()
```

Since the target tyoe is float, it' numerical and the task is regression.

### Step 3. Identify the number of samples and columns of both the data matrix and the target

```python
df.info()
```

### Step 4. Summarize the distribution of the data (min, max, median, mean, and standard deviation)

```python
df.aggregate(['min','max','median','mean','std'])
```

### Step 5. Plot the feature vs the target

```python
df.plot.scatter(x='Hours_Studied', y= 'Marks')
```

### Step 6. What is the relationship between the feature and the target? (increasing or decreasing or none)

Increasing

### Step 7. Initialize a `LinearRegression` model

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
```

### Step 8. Split the dataset into train and test sets

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x, y,
    test_size=0.20,
    random_state=42,
)
```

### Step 9.a Train the model on the training set

```python
model.fit(x_train.to_frame(), y_train)
```

### Step 9.b Identify the learned slope and the y-intercept of the model

```python
slope = model.coef_[0].round(2)
intercept = model.intercept_.round(2)

print(f'slope = {slope}\nintercept = {intercept}')
```

### Step 9.c how much marks do we get per 1 hour of study?

5.16

### Step 9.d how much does a 10 minutes increase of study time adds to our marks?

```python
print(f'10 minutes increases:{slope / 6}')
```

### Step 9.e if one doesn't study at all, how much marks do they get?

```python
print(intercept)
```

### Step 10. Evaluate the model on the test set

```python
y_predict = model.predict(x_test.to_frame())
```

```python
from sklearn.metrics import mean_absolute_error
error = mean_absolute_error(y_test, y_predict)
print(f'mean absolute error:{error}')
```

### Step 11. Take a sample of three rows from the test set and make a prediction on them and check whether they match the true label or not

```python
compared = pd.DataFrame(
    {
        'Actual Result': y_test.head(3),
        'Predicted': y_predict[:3]
    }
)
compared
```

### Step 12. Plot the data and the regression line

```python
fig = df.plot.scatter(
    x = 'Hours_Studied',
    y = 'Marks'
)
fig.plot(x_test, y_predict, color = 'red')
```
